```python
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.orm import declarative_base, sessionmaker
```

- create_engine: It is used to make connection from the database
- Column: It defines columns in database
- Integer and String: Datatype of values in database

```python
engine = create_engine('sqlite:///user.db')
```

```python
# this line is needed to create the table
Base = declarative_base()
```

```python
class User(Base):
    __tablename__ = 'users'

    id = Column(Integer, primary_key = True)
    name = Column(String)
    password = Column(String)
```

```python
Base.metadata.create_all(engine)
```

```python
Session = sessionmaker(bind = engine) # this line tells the sessionmaker to use the engine we created 
session = Session() # this line creates a session object
```

```python
# add a new user
new_user = User(name = 'Deepa', password = 'Deepa@8')
session.add(new_user)
session.commit()
```

```python
# add a new user
new_user = User(name = 'somu', password = 'somu@1')
session.add(new_user)
session.commit()
```

```python
# retrieve all users
users = session.query(User).all()

for user in users:
    print(f'User id: {user.id}, name: {user.name}, password: {user.password}')
```

```python
# update the user
user = session.query(User).filter_by(id=1).first() # first is used to get a single result
user.name = 'Deepanshi'
session.commit()

```

```python
# retrieve all users
users = session.query(User).all()
for user in users:
    print(f'User id: {user.id}, name: {user.name}, password: {user.password}')
```
